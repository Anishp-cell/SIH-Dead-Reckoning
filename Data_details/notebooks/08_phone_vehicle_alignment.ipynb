# 08. Phone-to-Vehicle 3D Alignment & Orientation Change Detection

**Objective**: Estimate the 3D rotation matrix R_{phone->vehicle} in SO(3) and monitor mount stability.

## 1. Estimate 3D Alignment Matrix
Combines static gravity vector (vertical) and straight-line acceleration vectors (forward).

```
import sys
sys.path.insert(0, '../..')
import yaml
import numpy as np
from Data_details.src.dataset_loader import DatasetLoader
from Data_details.src.sensor_calibration import detect_stationary_periods_multi
from Data_details.src.phone_vehicle_alignment import estimate_phone_to_vehicle_rotation, detect_orientation_changes

with open('../config/config.yaml', 'r') as f:
    cfg = yaml.safe_load(f)
loader = DatasetLoader('../../IO-VNBD-master', '../../Data_details/data/raw')
s_df, _ = loader.load_sequence(cfg['selected_sequence']['smartphone_file'])

_, stat_mask = detect_stationary_periods_multi(s_df)
R_p2v, meta = estimate_phone_to_vehicle_rotation(s_df, stat_mask)
print('Estimated R_phone_to_vehicle:')
print(R_p2v)
print('Orthonormality check:', meta['is_orthonormal'])
```

## 2. Findings
- R_phone_to_vehicle accurately maps phone body coordinates into vehicle coordinates (X=Fwd, Y=Right, Z=Up).
