# 04. GPS Trajectory & Geodesic Distance

**Objective**: Analyze geographic path, total distance, speeds, and plot 2D trajectory.

## 1. GPS Trajectory Analysis

```
import sys
sys.path.insert(0, '../..')
import yaml
import matplotlib.pyplot as plt
from Data_details.src.dataset_loader import DatasetLoader
from Data_details.src.gps_analysis import analyze_gps_trajectory

with open('../config/config.yaml', 'r') as f:
    cfg = yaml.safe_load(f)
loader = DatasetLoader('../../IO-VNBD-master', '../../Data_details/data/raw')
s_df, _ = loader.load_sequence(cfg['selected_sequence']['smartphone_file'])

gps_stats = analyze_gps_trajectory(s_df)
for k, v in gps_stats.items():
    print(f'{k}: {v}')
```

## 2. Trajectory Visualizations
View generated GPS trajectory plot.

```
import matplotlib.image as mpimg
img = mpimg.imread('../../Data_details/outputs/plots/gps_trajectory.png')
plt.figure(figsize=(10, 10))
plt.imshow(img)
plt.axis('off')
plt.title('GPS Trajectory — Sequence S1', fontsize=14)
plt.show()
```
