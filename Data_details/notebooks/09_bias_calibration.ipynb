# 09. Multi-Sensor Stationary Detection & Bias Calibration

**Objective**: Detect stopped intervals and quantify zero-rate gyroscope bias and accelerometer offset.

## 1. Multi-Sensor Stationary Detection

```
import sys
sys.path.insert(0, '../..')
import yaml
import pandas as pd
from Data_details.src.dataset_loader import DatasetLoader
from Data_details.src.sensor_calibration import detect_stationary_periods_multi, estimate_gyroscope_bias, estimate_accelerometer_bias

with open('../config/config.yaml', 'r') as f:
    cfg = yaml.safe_load(f)
loader = DatasetLoader('../../IO-VNBD-master', '../../Data_details/data/raw')
s_df, _ = loader.load_sequence(cfg['selected_sequence']['smartphone_file'])

periods, stat_mask = detect_stationary_periods_multi(s_df)
print(f'Detected {len(periods)} multi-sensor stationary windows.')
gyro_b = estimate_gyroscope_bias(s_df, stat_mask)
accel_b = estimate_accelerometer_bias(s_df, stat_mask)
print('Gyroscope Bias:', gyro_b)
print('Accelerometer Bias:', accel_b)
```

## 2. Findings
- Gyroscope zero-rate bias vector (|b_g| ~ 0.005 rad/s) is stable across stops and directly subtractable.
