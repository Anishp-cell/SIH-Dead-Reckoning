# 10. World-Frame Gravity Compensation

**Objective**: Rotate body acceleration to Navigation ENU frame and subtract [0, 0, g]^T to eliminate gravity leakage.

## 1. Run Gravity Compensation

```
import sys
sys.path.insert(0, '../..')
import yaml
import numpy as np
from Data_details.src.dataset_loader import DatasetLoader
from Data_details.src.attitude_estimation import estimate_sequence_attitude
from Data_details.src.gravity_compensation import compensate_gravity_world_frame, evaluate_gravity_compensation
from Data_details.src.sensor_calibration import detect_stationary_periods_multi

with open('../config/config.yaml', 'r') as f:
    cfg = yaml.safe_load(f)
loader = DatasetLoader('../../IO-VNBD-master', '../../Data_details/data/raw')
s_df, _ = loader.load_sequence(cfg['selected_sequence']['smartphone_file'])

quats = estimate_sequence_attitude(s_df)
a_dyn, _ = compensate_gravity_world_frame(s_df[['acc_x', 'acc_y', 'acc_z']].values, quats)
_, stat_mask = detect_stationary_periods_multi(s_df)
res = evaluate_gravity_compensation(s_df, a_dyn, stat_mask)
print('Gravity Compensation Evaluation:', res)
```

## 2. Findings
- Orientation-aware gravity compensation reduces stationary residual acceleration, eliminating the dominant cause of quadratic position error.
