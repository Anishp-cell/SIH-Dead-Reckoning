# 14. Vehicle Vibration Decomposition & Road Roughness Analysis

**Objective**: Decompose raw signals into kinematic translation and structural vibration, comparing stationary vs cruising states.

## 1. Setup & Vibration Decomposition

```
import sys
sys.path.insert(0, '../..')
import numpy as np
import matplotlib.pyplot as plt
from Data_details.src.vibration_analysis import decompose_vibration, compute_vibration_metrics
from Data_details.src.dataset_loader import DatasetLoader
import yaml

with open('../config/config.yaml', 'r') as f:
    cfg = yaml.safe_load(f)
loader = DatasetLoader('../../IO-VNBD-master', '../../Data_details/data/raw')
s_df, _ = loader.load_sequence(cfg['selected_sequence']['smartphone_file'])
az = s_df['acc_z'].values
kin, vib = decompose_vibration(az, cutoff_hz=1.8, fs=10.0)
metrics = compute_vibration_metrics(az, vib)
print('Vertical Vibration Metrics:', metrics)
```

## 2. Visualize Isolated Vibration

```
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(vib[2000:2500], color='#e74c3c', lw=0.8)
ax.set_title('Isolated Road/Engine Vibration (>= 1.8 Hz)')
ax.set_ylabel('Vibration (m/s²)')
ax.grid(True, alpha=0.3)
plt.show()
```
