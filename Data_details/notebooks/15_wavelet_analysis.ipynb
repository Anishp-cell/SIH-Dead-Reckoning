# 15. Wavelet Multiresolution Denoising & Non-Stationary Transient Analysis

**Objective**: Apply Discrete Wavelet Transform (DWT) multi-level thresholding and compare against linear Butterworth filtering.

## 1. Setup & Wavelet Denoising

```
import sys
sys.path.insert(0, '../..')
import numpy as np
import matplotlib.pyplot as plt
from Data_details.src.wavelet_denoising import wavelet_denoise_signal
from Data_details.src.filter_design import apply_butterworth_causal
from Data_details.src.dataset_loader import DatasetLoader
import yaml

with open('../config/config.yaml', 'r') as f:
    cfg = yaml.safe_load(f)
loader = DatasetLoader('../../IO-VNBD-master', '../../Data_details/data/raw')
s_df, _ = loader.load_sequence(cfg['selected_sequence']['smartphone_file'])
ay = s_df['acc_y'].values
wav_clean, wmeta = wavelet_denoise_signal(ay, wavelet='sym4', level=3)
print('Wavelet Denoising Meta:', wmeta)
```

## 2. Wavelet vs Butterworth Residuals

```
butter_clean = apply_butterworth_causal(ay, cutoff_hz=1.5, fs=10.0)
plt.figure(figsize=(12, 5))
plt.plot(ay[500:800], color='#bdc3c7', label='Raw Signal')
plt.plot(butter_clean[500:800], 'r--', label='Butterworth (1.5 Hz)')
plt.plot(wav_clean[500:800], 'b-', label='Wavelet (sym4, Level 3)')
plt.title('Wavelet vs Butterworth Filter on Transient Deceleration')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
