# 05. WGS84 Geodetic to Local Metric ENU Coordinate Transformation

**Objective**: Transform spherical GPS coordinates into local Cartesian East-North-Up tangent plane.

## 1. Run Geodetic to ENU Conversion

```
import sys
sys.path.insert(0, '../..')
import yaml
import numpy as np
from Data_details.src.dataset_loader import DatasetLoader
from Data_details.src.coordinate_transform import geodetic_to_enu

with open('../config/config.yaml', 'r') as f:
    cfg = yaml.safe_load(f)
loader = DatasetLoader('../../IO-VNBD-master', '../../Data_details/data/raw')
s_df, _ = loader.load_sequence(cfg['selected_sequence']['smartphone_file'])

lat = s_df['gps_lat'].values
lon = s_df['gps_lon'].values
alt = s_df['gps_alt'].values

east, north, up, origin = geodetic_to_enu(lat, lon, alt)
print('ENU Origin:', origin)
print(f'East range: [{np.nanmin(east):.1f}m, {np.nanmax(east):.1f}m]')
print(f'North range: [{np.nanmin(north):.1f}m, {np.nanmax(north):.1f}m]')
```

## 2. Local ENU Trajectory Plot

```
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
img = mpimg.imread('../../Data_details/outputs/plots/enu_trajectory.png')
plt.figure(figsize=(10, 10))
plt.imshow(img)
plt.axis('off')
plt.show()
```

- WGS84 coordinates successfully mapped to metric East, North, Up space centered at (52.401660°, -1.505290°).
- Local Cartesian coordinates allow direct integration and Euclidean error measurement without spherical distortion.
